```python
movies=df[
    [
    'title',
    'genres',
    'overview',
    'release_date',
    'vote_average',
    'vote_count',
    'original_language',
    'popularity'
    ]
].copy()

movies.head()
```

```python
movies = movies.dropna(subset=['overview']).copy()

print("Dataset shape:", movies.shape)
print("Missing values:")
print(movies.isnull().sum())
```

```python
movies['genres'].head(5)
```

```python
import ast

def get_genres(genre_text):
    genre_list = ast.literal_eval(genre_text)
    return [genre['name'] for genre in genre_list]

movies['genres'] = movies['genres'].apply(get_genres)

movies[['title', 'genres']].head(5)
```

```python
all_genres = sorted(set(
    genre
    for genre_list in movies['genres']
    for genre in genre_list
))

print(all_genres)
print("Total genres:", len(all_genres))
```

```python
print(all_genres)
print("Total genres:", len(all_genres))
```

```python
def recommend_by_genre(genre, n=10):
    recommendations = movies[
        movies['genres'].apply(lambda genres: genre in genres)
    ]

    recommendations = recommendations.sort_values(
        by=['vote_count', 'popularity'],
        ascending=False
    )

    return recommendations[
        ['title', 'genres', 'overview', 'release_date', 'vote_average']
    ].head(n)
```

```python
recommend_by_genre("Romance")
```

```python
recommend_by_genre("Romance", 5)
```

```python
movies['overview'] = movies['overview'].fillna('')

movies['overview_clean'] = (
    movies['overview']
    .str.lower()
    .str.strip()
)

movies[['title', 'overview_clean']].head(5)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("TF-IDF and cosine similarity are ready!")
```

```python
tfidf = TfidfVectorizer(stop_words='english', max_features=5000)

tfidf_matrix = tfidf.fit_transform(movies['overview_clean'])

print("Matrix shape:", tfidf_matrix.shape)
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
def recommend_similar_movies(title, n=10):
    matches = movies[
        movies['title'].str.casefold() == title.casefold()
    ]

    if matches.empty:
        print("Movie not found. Check the title spelling.")
        return

    position = movies.index.get_loc(matches.index[0])

    scores = list(enumerate(similarity_matrix[position]))
    scores = sorted(scores, key=lambda item: item[1], reverse=True)

    scores = [
        item for item in scores if item[0] != position
    ][:n]

    positions = [item[0] for item in scores]

    recommendations = movies.iloc[positions][
        ['title', 'genres', 'overview', 'release_date', 'vote_average']
    ].copy()

    recommendations['similarity_score'] = [
        round(score, 3) for _, score in scores
    ]

    return recommendations
```

```python
recommend_similar_movies("The Dark Knight")
```

```python
def recommend_by_genre_and_movie(genre, favorite_movie, n=10):
    matches = movies[
        movies['title'].str.casefold()
        == favorite_movie.casefold()
    ]

    if matches.empty:
        print("Favorite movie not found. Check the title spelling.")
        return

    position = movies.index.get_loc(matches.index[0])

    genre_matches = movies['genres'].apply(
        lambda genres: genre in genres
    )

    candidates = movies[genre_matches].copy()
    candidates['similarity_score'] = similarity_matrix[
        position, [movies.index.get_loc(i) for i in candidates.index]
    ]

    candidates = candidates[
        candidates.index != movies.index[position]
    ]

    return candidates.sort_values(
        by=['similarity_score', 'vote_count'],
        ascending=False
    )[
        ['title', 'genres', 'overview', 'release_date',
         'vote_average', 'similarity_score']
    ].head(n)
```

```python
recommend_by_genre_and_movie(
    "Action",
    "The Dark Knight",
    10
)
```

```python
def find_movie(title, year):
    results = movies[
        (movies['title'].str.casefold() == title.casefold()) &
        (pd.to_datetime(
            movies['release_date'], errors='coerce'
        ).dt.year == year)
    ]

    if results.empty:
        print("No matching movie found.")
    else:
        return results[
            ['title', 'release_date', 'original_language',
             'genres', 'vote_average']
        ]
```

```python
find_movie("Batman", 1989)
```

```python
def recommend_by_language(language, n=10):
    results = movies[
        movies['original_language'].str.lower() == language.lower()
    ].copy()

    results = results.sort_values(
        by=['vote_count', 'popularity'],
        ascending=False
    )

    return results[
        ['title', 'original_language', 'genres',
         'release_date', 'vote_average']
    ].head(n)
```

```python
recommend_by_language("ko", 10)
```

```python
def recommend_by_rating(min_rating=7.0, n=10):
    results = movies[
        movies['vote_average'] >= min_rating
    ].copy()

    results = results.sort_values(
        by=['vote_average', 'vote_count'],
        ascending=False
    )

    return results[
        ['title', 'genres', 'release_date',
         'vote_average', 'vote_count']
    ].head(n)
```

```python
recommend_by_rating(7.5, 10)
```

```python
mood_genres = {
    "happy": ["Comedy", "Animation", "Family"],
    "sad": ["Comedy", "Romance", "Family"],
    "romantic": ["Romance"],
    "adventurous": ["Adventure", "Action", "Fantasy"],
    "scared": ["Horror", "Thriller"],
    "curious": ["Mystery", "Science Fiction", "Crime"],
    "inspired": ["Drama", "History"],
    "relaxed": ["Comedy", "Family", "Music"]
}

def recommend_by_mood(mood, n=10):
    mood = mood.lower().strip()

    if mood not in mood_genres:
        print("Mood not found. Available moods:", list(mood_genres))
        return

    selected_genres = mood_genres[mood]

    results = movies[
        movies['genres'].apply(
            lambda genres: any(
                genre in genres for genre in selected_genres
            )
        )
    ].copy()

    results = results.sort_values(
        by=['vote_count', 'popularity'],
        ascending=False
    )

    return results[
        ['title', 'genres', 'overview',
         'release_date', 'vote_average']
    ].head(n)
```

```python
recommend_by_mood("sad", 10)
```

```python
def show_movies(recommendations):
    if recommendations is None or recommendations.empty:
        print("No movies found.")
        return

    for _, movie in recommendations.iterrows():
        print("=" * 60)
        print("Title:", movie['title'])
        print("Genres:", ", ".join(movie['genres']))
        print("Rating:", movie['vote_average'])
        print("Release date:", movie['release_date'])
        print("Overview:", movie['overview'])
        print()
```

```python
results = recommend_by_mood("romantic", 5)
show_movies(results)
```
